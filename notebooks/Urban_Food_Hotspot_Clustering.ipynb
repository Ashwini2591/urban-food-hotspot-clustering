```python
import numpy as np
import pandas as pd
from sklearn.cluster import DBSCAN

# Use data with 'lat' and 'lon' (ratings_locations_cleaned)
X = ratings_locations_cleaned[['lat', 'lon']].dropna().values

best_labels = None
best_clusters = -1 # Initialize to -1 to ensure best_eps is always set if any clusters are found
best_eps = np.nan # Initialize best_eps to handle cases where no clusters are found

# Try multiple eps values automatically
for eps in np.linspace(0.002, 0.02, 10):
    db = DBSCAN(eps=eps, min_samples=5).fit(X)
    labels = db.labels_

    # count clusters (ignore -1 noise)
    n_clusters = len(set(labels)) - (1 if -1 in labels else 0)

    if n_clusters > best_clusters:
        best_clusters = n_clusters
        best_labels = labels
        best_eps = eps

# Assign best result to the cleaned ratings locations dataframe
# Ensure you make a copy to avoid SettingWithCopyWarning
df_clustered = ratings_locations_cleaned.copy()

if best_labels is not None:
    df_clustered['DBSCAN_cluster'] = best_labels
else:
    # If no clusters were found, assign -1 to all, indicating noise/no cluster
    df_clustered['DBSCAN_cluster'] = -1

print("Best eps:", best_eps)
print("Number of clusters:", best_clusters if best_clusters != -1 else 0)
print(df_clustered['DBSCAN_cluster'].value_counts())
```

```python
def plot_dbscan_clusters(data):

    m = folium.Map(location=[12.97, 77.59], zoom_start=12)

    clustered = data[data['DBSCAN_cluster'] != -1]

    if clustered.empty:
        print("⚠️ No clusters found. Increase eps further.")
        return m

    cluster_centers = clustered.groupby('DBSCAN_cluster').agg({
        'lat': 'mean',
        'lon': 'mean',
        'DBSCAN_cluster': 'count'
    }).rename(columns={'DBSCAN_cluster': 'size'}).reset_index()

    for _, row in cluster_centers.iterrows():

        folium.Circle(
            location=[row['lat'], row['lon']],
            radius=row['size'] * 50,   # 🔥 much bigger
            color='purple',
            fill=True,
            fill_color='purple',
            fill_opacity=0.6,
            popup=f"Cluster size: {row['size']}"
        ).add_to(m)

    return m

map_dbscan = plot_dbscan_clusters(ratings_locations_cleaned)
map_dbscan
```
